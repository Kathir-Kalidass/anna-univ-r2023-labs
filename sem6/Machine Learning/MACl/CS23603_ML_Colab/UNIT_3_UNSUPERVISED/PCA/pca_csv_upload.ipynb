Apply PCA on uploaded CSV using numeric columns and visualize principal components.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from google.colab import files
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

uploaded = files.upload()
file_name = next(iter(uploaded))
df = pd.read_csv(file_name)
display(df.head())

numeric_df = df.select_dtypes(include=np.number).dropna()
if numeric_df.shape[1] < 2:
    raise ValueError("Need at least two numeric columns for PCA visualization.")

X_scaled = StandardScaler().fit_transform(numeric_df)
pca = PCA(n_components=2, random_state=42)
X_pca = pca.fit_transform(X_scaled)

plt.figure(figsize=(7, 5))
plt.scatter(X_pca[:, 0], X_pca[:, 1], alpha=0.7)
plt.title("PCA Projection of Uploaded CSV")
plt.xlabel("PC1")
plt.ylabel("PC2")
plt.show()

print("Explained variance ratio:", pca.explained_variance_ratio_)
print("Cumulative explained variance:", np.cumsum(pca.explained_variance_ratio_))
```
