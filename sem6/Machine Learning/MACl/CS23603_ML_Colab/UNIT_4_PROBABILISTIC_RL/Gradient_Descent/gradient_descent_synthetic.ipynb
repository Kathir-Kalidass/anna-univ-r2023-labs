```python
import numpy as np
import matplotlib.pyplot as plt

# Part A: Minimize f(x) = x^2
def f(x):
    return x ** 2

def grad(x):
    return 2 * x

x = 5.0
lr = 0.1
history = []
for _ in range(25):
    x = x - lr * grad(x)
    history.append(x)

x_vals = np.linspace(-6, 6, 200)
plt.figure(figsize=(7, 4))
plt.plot(x_vals, f(x_vals))
plt.scatter(history, [f(v) for v in history], color="red", s=20)
plt.title("Gradient Descent on f(x)=x^2")
plt.show()

# Part B: Linear regression with manual GD
np.random.seed(42)
X = 2 * np.random.rand(120)
y = 4 + 3 * X + np.random.randn(120)

w, b = 0.0, 0.0
epochs = 150
lr = 0.1
n = len(X)
costs = []

for _ in range(epochs):
    y_pred = w * X + b
    dw = (-2 / n) * np.sum(X * (y - y_pred))
    db = (-2 / n) * np.sum(y - y_pred)
    w -= lr * dw
    b -= lr * db
    costs.append(np.mean((y - y_pred) ** 2))

plt.figure(figsize=(7, 4))
plt.plot(costs)
plt.title("Cost vs Iterations")
plt.xlabel("Iteration")
plt.ylabel("MSE")
plt.show()

print("Final parameters -> w:", w, "b:", b)
```
